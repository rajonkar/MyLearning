```python
from IPython.display import display, Markdown 

# Access the notebook's cells in memory
try:
    shell = get_ipython()
    cells = shell.user_ns['In']  # the input history list
except:
    cells = []

# We can only assign to nbformat cells if we read the notebook
# So let's do a safe assign: only assign ID if it doesn't exist
import nbformat
import os

notebook_path = '01_05_JupyterNotebook.ipynb'  # Replace with your notebook filename
if not os.path.exists(notebook_path):
    raise FileNotFoundError(f"Notebook file '{notebook_path}' not found.")

# Load notebook from file
nb = nbformat.read(notebook_path, as_version=4)

# Find the max existing numeric ID
max_id = 0
for cell in nb.cells:
    pid = cell.metadata.get('persistent_id')
    if pid and pid.isdigit():
        max_id = max(max_id, int(pid))

# Assign numeric IDs ONLY to cells missing them
new_ids = 0
for cell in nb.cells:
    if 'persistent_id' not in cell.metadata:
        max_id += 1
        cell.metadata['persistent_id'] = str(max_id)
        new_ids += 1

# Save notebook back
nbformat.write(nb, notebook_path)

display(Markdown(f"✅ {new_ids} new numeric IDs assigned. Max ID: {max_id}"))
```
